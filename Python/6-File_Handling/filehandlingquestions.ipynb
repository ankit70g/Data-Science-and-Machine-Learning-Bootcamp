```python
def read_file():
    with open("sample.txt", "r") as file:
        for line in file:
            print(line.strip())
read_file()
```

```python
def write_file():
    data = ["Ankit", "Shanaya", "Kirti"]
    with open("output.txt", "w") as file:
        for item in data:
            file.write(item + "\n")

write_file()
        
```

```python
def copy_content():
    with open("output.txt", "r") as out:
        content = out.read()

    with open("dest.txt", "w") as dest:
        dest.write(content)

copy_content()
```

```python
def string_app():
    with open("dest.txt", "a") as file:
        file.write("\nGood Night")

string_app()
```

```python
def count_words():
    with open("dest.txt", "r") as file:
        text = file.read()
        words = text.split()
        print(len(words))

count_words()
```

```python
def find_replace():
    with open("output.txt", "r") as file:
        content = file.read()

    content = content.replace("Ankit", "Yash")

    with open("dest.txt", "w") as file:
        file.write(content)

find_replace()
```

```python
def reverse_read():
    with open("dest.txt", "r") as file:
        lines = file.readlines()

    for line in reversed(lines):
        print(line.strip())

reverse_read()
```

```python
def count_stats():
    with open("dest.txt", "r") as file:
        content = file.read()

    lines = content.split("\n")
    words = content.split()
    chars = len(content)

    print("Lines:", len(lines))
    print("Words:", len(words))
    print("Characters:", chars)

count_stats()
```

```python
def merged_files():
    files = ["file1.txt", "file2.txt"]

    with open("merged.txt", "w") as outfile:
        for fname in files:
            with open(fname, "r") as infile:
                outfile.write(infile.read() + "\n")

merged_files()
```

```python
def split_file():
    with open("large.txt", "r") as file:
        content = file.read()

    lines = content.split(" ")
    
    for i in range(0, len(lines), 100):
        with open(f"part_{i//100}.txt", "w") as f:
            f.writelines(lines[i:i+100])

split_file()
```

```python
from datetime import datetime

def create_log():
    with open("activity.log", "a") as file:
        file.write(f"{datetime.now()} - User logged in\n")

create_log()
```

```python
def binary_copy():
    with open("image.bin", "rb") as src:
        data = src.read()

    with open("copy_image.bin", "wb") as dest:
        dest.write(data)

binary_copy()
```

```python
import csv

def csv_read():
    with open("data.csv", "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            print(row)

csv_read()
```

```python
import json

def read_json():
    with open("data.json", "r") as file:
        data = json.load(file)
        print(data)

read_json()
```

```python
def read_protected():
    try:
        with open("protected.txt", "r") as file:
            print(file.read())
    except PermissionError:
        print("Permission denied!")
    except FileNotFoundError:
        print("File not found!")

read_protected()
```
