# File Operation - Read and Write Files

1. Introduction

    File handling refers to the process of performing operations on a file, such as creating, opening, reading, writing and closing it through a programming interface. It involves managing the data flow between the program and the file system on the storage device, ensuring that data is handled safely and efficiently.

2. Opening a file

    To open a file, we can use open() function, which requires file-path and mode as arguments.

    Syntax: file = open('filename.txt', 'mode')
    - filename.txt: name (or path) of the file to be opened.
    - mode: mode in which you want to open the file (read, write, append, etc.).

```python
f = open("example.txt", "r")
print(f)
```

3. Closing a File

    The file.close() method closes the file and releases the system resources. If the file was opened in write or append mode, closing ensures that all changes are properly saved.

```python
file = open("example.txt", "r")
# Perform file operations
file.close()
```

- Handling exceptions when closing a File

```python
try:
    file = open("example.txt", "r")
    content = file.read()
    print(content)
finally:
    file.close()
```

4. Checking File Properties

    - f.name: Returns the name of the file that was opened (in this case, "example.txt").
    - f.mode: Tells us the mode in which the file was opened. Here, it’s 'r' which means read mode.
    - f.closed: Returns a boolean value- False when file is currently open otherwise True.

```python
f = open("example.txt", "r")
print("Filename:", f.name)
print("Mode:", f.mode)
print("Is Closed?", f.closed)

f.close()
print("Is Closed?", f.closed)
```

5. Reading a File

    Reading a file can be achieved by file.read() which reads the entire content of the file. After reading, it’s good practice to close the file to free up system resources.

- "r": 	Read-only. Raises I/O error if file doesn't exist.

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

- "r+": Read and write. Raises I/O error if the file does not exist.

```python
with open('example.txt', 'r+') as file:
    content = file.read()
    file.write('\nThis is a new line.')
```

- Read a file line by line

```python
with open("example.txt", "r") as file:
    for line in file:
        print(line)
```

- Remove the new line character using strip()

```python
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())
```

6. Writing a File

    In Python, writing to a file is done using the mode "w". This creates a new file if it doesn’t exist, or overwrites the existing file if it does. The write() method is used to add content. After writing, make sure to close the file.

- Writing a file (Overwriting)

    ‘w’	Write-only. Overwrites file if it exists, else creates a new one.

```python
with open("example.txt", "w") as file:
    file.write("Hello World\n")
    file.write("File handling is easy with Python")

print("File written successfully")
```

- ‘w+’:	Read and write. Overwrites file or creates new one.

```python
with open('example.txt', 'w+') as file:
    file.write('Hello, world!')
    file.seek(0) #Move the cursor to the beginning
    content = file.read()
```

- Writing a list of lines to a file

```python
lines = ["First line \n", "Second line \n", "Third line \n"]
with open("example.txt", "a") as file:
    file.writelines(lines)
print("Successfully wrote a list of lines")
```

7. Updating a file

- ‘a’: Append-only. Adds data to end. Creates file if it doesn't exist.

```python
with open("example.txt", "a") as file:
    file.write("Append operation taking place!\n")

print("Successfully updated the file")
```

- 'a+': Read and append. Pointer at end. Creates file if it doesn't exist.

```python
with open("example.txt", "a+") as file:
    file.write("Append+ operation taking place\n")

print("Successfully updated the file")
```

8. Working with Binary Files

- "wb": Write in binary. Overwrites or creates new.

```python
data = b"\x00\x01\x02\x03\x04"
with open("example.bin", "wb") as file:
    file.write(data)

print("Successfully wrote to a binary file")
```

- ‘rb’:	Read in binary mode. File must exist.

```python
with open("example.bin", "rb") as file:
    content = file.read()
    print(content)

print("Successfully wrote to a binary file")
```

- "ab": Adds data at the end of file (does NOT overwrite)

```python
with open("example.bin", "ab") as file:
    file.write(b"!!!")
```

- "rb+": Allows both reading and writing (no overwrite by default)

```python
with open("example.bin", "rb+") as file:
    print(file.read())

    file.write(b" NEW")
```

- "wb+": Overwrites file and allows read/write

```python
with open("example.bin", "wb+") as file:
    file.write(b"Fresh Data")

    file.seek(0)
    print(file.read())
```

- "ab+": Append + also read

```python
with open("example.bin", "ab+") as file:
    file.write(b" Added")

    file.seek(0)
    print(file.read())
```

9. Practice examples

- Reading the content from a source text file and writing to a destination text file

```python
with open("example.txt", "r") as source_file:
    content = source_file.read()

with open("destination.txt", "w") as destination_file:
    destination_file.write(content)
```

- Read a text file and count the number of lines, words, and characters

```python
def count_text_file(file_path):
    with open(file_path, "r") as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = "example.txt"
lines, words, characters = count_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```
