1. Introduction to map() function

    map() is a built-in function used to apply a function to every item in an iterable (like a list, tuple, etc.).

    Instead of writing loops, map() does the transformation automatically.


2. Syntax

```python
map(function, iterable)
```

3. Creating a map() function

```python
def square(x):
    return x*x

square(4)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7]
list(map(square, numbers))
```

4. Converting map object to a list

    By default, map() function returns a map object, which is an iterator. In many cases, we will need to convert this iterator to a list to work with the results directly.

```python
def double(val):
    return val * 2

a = [1, 2, 3, 4, 5]
res = list(map(double, a))
print(res)
```

5. map() with lambda

```python
numbers = [1, 2, 3, 4, 5]
list(map(lambda x: x*x, numbers))
```

6. map() with multiple iterables

```python
num1 = [1, 2, 3]
num2 = [4, 5, 6]

added_numbers = list(map(lambda x, y: x + y, num1, num2))
print(added_numbers)
```

7. Converting a list of string to integers using map()

```python
str_num = ["1", "2", "3", "4", "5"]
int_num = list(map(int, str_num))

print(int_num)
```

8. Converting strings to Uppercase

```python
fruits = ["apple", "litchi", "orange"]
upper_case = list(map(str.upper, fruits))

print(upper_case)
```

9. Applying a map() to a list of dictionaries

```python
def get_name(person):
    return person["name"]

people = [
    {"name": "Ankit", "age": 19},
    {"name": "Shanaya", "age": 21},
    {"name": "Yash", "age": 32},
]

print(list(map(get_name, people)))
```

10. Extracting first character from strings

```python
words = ["apple", "litchi", "banana"]
res = map(lambda s: s[0], words)

print(list(res))
```

11. Removing whitespaces from strings

```python
words = ["  hello", "Ankit ", " Gupta "]
res = map(str.strip, words)
print(list(res))
```
