```python
def fib_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]

print(fib_memo(10))
print(fib_memo(50))

```

```python
def update_dic(a, b=None):
    if b is None:
        b = {}

    b[a] = a * 10
    return b

print(update_dic(1))
print(update_dic(2, {"x": 100}))
```

```python
def filter_ints(**kwargs):
    return {k: v for k, v in kwargs.items() if isinstance(v, int)}

print(filter_ints(a = 10, b = "hello", c = 25, d = 3.5))
```

```python
def apply_callback(callback, numbers):
    return [callback(x) for x in numbers]

def square(x):
    return x*x

def cube(x):
    return x**3

print(apply_callback(square, [1, 2, 3]))
print(apply_callback(cube, [1, 2, 3]))
```

```python
def square_fn():
    def square(x):
        return x * x
    return square

sq = square_fn()

print(sq(12))
```

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Execution Time:", end - start)
        return result
    return wrapper

@timer
def complex_calculation(n):
    total = 0
    for i in range(n):
        total += i**2
    return total

print(complex_calculation(100000))
```

```python
def filter_and_map(filter_fn, map_fn, numbers):
    filtered = filter(filter_fn, numbers)
    return list(map(map_fn, filtered))

print(filter_and_map(lambda x: x%2==0,
                     lambda x: x*x,
                     [1, 2, 3, 4, 5, 6]))
```

```python
def compose(f, g):
    return lambda x: f(g(x))

def add2(x):
    return x + 2

def multiply3(x):
    return x * 3

result = compose(add2, multiply3)

print(result(5))
```

```python
from functools import partial

def multiply(a, b):
    return a * b

double = partial(multiply, 2)

print(double(5))
print(double(10))
```

```python
def average(numbers):
    try:
        return sum(numbers) / len(numbers)
    except Exception:
        return None

print(average([10, 20 , 30]))
print(average([]))
```

```python
def fib_gen():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

g = fib_gen()
print([next(g) for _ in range(10)])
```

```python
def curry_prod(a):
    def second(b):
        def third(c):
            return a*b*c
        return third
    return second

print(curry_prod(2)(3)(4))
```

```python
def write_numbers(filename, numbers):
    try:
        with open(filename, "w") as f:
            for n in numbers:
                f.write(str(n) + "\n")
        print("File written successfully")
    except Exception as e:
        print("Error:", e)

write_numbers("numbers.txt", [1,2,3,4,5])
```

```python
def sort_types(mixed):
    return (
        [x for x in mixed if type(x) == int],
        [x for x in mixed if type(x) == str],
        [x for x in mixed if type(x) == float]
    )

print(sort_types(["Ankit", 1, 32.1, "Gupta", 2, 3]))
```
