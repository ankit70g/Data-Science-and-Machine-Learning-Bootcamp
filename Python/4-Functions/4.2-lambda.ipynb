1. Introduction to Lambda functions

    Lambda functions are small anonymous functions defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

2. Syntax

```python
# Syntax
function_name = lambda arguments: expression
```

```python
# Basic function
def addition(a, b):
    return a + b

addition(2, 4)
```

3. Creating a Lambda Function

```python
addition = lambda a,b: a+b
type(addition)
print(addition(5, 6))
```

```python
def even(num):
    if num % 2 == 0:
        return True
    
even(20)
```

```python
even = lambda num: num % 2 == 0
print(even(14))
```

```python
def addition(x, y, z):
    return x+y+z

addition(12,13,14)
```

```python
addition = lambda x,y,z: x+y+z
addition(12,13,14)
```

4. Use Cases of Lambda Functions

<div style="margin-left:40px">
a. Using with Condition Checking

```python
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"

print(check(5))
print(check(-3))
print(check(0))
```

<div style="margin-left:40px">
b. Using with List Comprehension

```python
func = [lambda arg=x: arg * 10 for x in range(1, 4)]
for i in func:
    print(i())
```

<div style="margin-left:40px">
c. Using for Returning Multiple Results

```python
calc = lambda x, y: (x+y,x*y)
res = calc(3, 4)
print(res)
```

<div style="margin-left:40px">
d. Using with filter()

filter() function uses a lambda expression to select elements from a list that satisfy a given condition, such as keeping only even numbers.

```python
n = [1, 2, 3, 4, 5]
even = filter(lambda x: x % 2 == 0, n)
print(list(even))
```

<div style="margin-left:40px">
e. Using with map()

map() function applies a lambda expression to each element of a list and returns a new list with the transformed values.

```python
a = [1, 2, 3, 4, 5]
double = map(lambda x: x * 2, a)
print(list(double))
```

<div style="margin-left:40px">
f. Using with reduce()

reduce() function repeatedly applies a lambda expression to elements of a list to combine them into a single result.

```python
from functools import reduce
a = [1,2,3,4,5]
mult = reduce(lambda x, y: x * y, a)
print(mult)
```
