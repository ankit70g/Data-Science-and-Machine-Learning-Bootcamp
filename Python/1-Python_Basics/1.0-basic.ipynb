Syntax and Semantics in Python

Syntax refers to the set of rules that defines the combinations of symbols that are considered to be correctly structured programs in a language. In simpler terms, syntax is about the correct arrangement of words and symbols in a code.


Basic syntax rules in Python
- Case sensitivity - Python is strictly case-sensitive. This means that uppercase and lowercase letters are treated as completely different characters.

```python
City = "London" 
city = "Paris" 
CITY = "Tokyo"
print(City)
print(city)
print(CITY)
```

- Indentation - Indentation in Python is used to define the structure and hierarchy of the code. In most programming languages, you use curly braces {} to group blocks of code together. In Python, you use indentation (spaces or tabs at the beginning of a line).

```python
# Correct Indentation
if 5 > 2:
    print("Five is greater than two!")  # This is indented
```

```python
# Incorrect Indentation (Will cause an error)
if 5 > 2:
print("Five is greater than two!")
```

- Comments - Comments are notes you leave in your code for yourself or other developers. Python completely ignores them when running the program.

```python
#Single-line comments: Start with a hash symbol #.
#This is a single line comment

#Multi-line comments: Enclosed within three quotation marks """ or '''.
"""
This is a multi-line comment.
It can span across several lines.
Great for writing detailed explanations.
"""
```

- Line Continuation - In Python, line continuation is how you tell the interpreter that a single statement is spanning across multiple lines.

```python
#Implicit Line Continuation (The Preferred Way) - Python automatically assumes a line continues if it is inside parentheses (), square brackets [], or curly braces {}

bootcamp_topics = [
    "Python Basics",
    "Data Analysis with Pandas",
    "Machine Learning",
    "Deep Learning"
]
print(bootcamp_topics)
```

```python
#Explicit Line Continuation (The Backslash) - If your code does not naturally use parentheses, brackets, or braces, you can force a line continuation by placing a backslash (\) at the very end of the line.
total=1+2+3+4+5+6+7+\
4+5+6

print(total)
```
