### Introduction to Operators
Python operators are special symbols or keywords used to perform operations on variables and values. They are categorized into several groups, each serving a specific purpose in building expressions and controlling program flow.

### Arithmetic Operators

```python
a = 10
b = 5

add_result = a + b # Addition
sub_result = a - b # Subtraction
mult_result = a * b # Multiplication
div_result = a / b  # Division (returns a float)
floor_div_result = a // b # Floor Division (returns the largest integer less than or equal to the division result)
mod_result = a % b  # Modulus (returns the remainder of the division)
exp_result = a ** b # Exponentiation (returns a raised to the power of b)

print("Addition:", add_result)
print("Subtraction:", sub_result)
print("Multiplication:", mult_result)
print("Division:", div_result)
print("Floor Division:", floor_div_result)
print("Modulus:", mod_result)
print("Exponentiation:", exp_result)
```

### Comparison Operators

```python
## ==

a = 10
b = 10

a==b
```

```python
str1 = "Ankit"
str2 = "ankit"

str1==str2
```

```python
## Not Equal To (!=)

str1 != str2
```

```python
## greater than (>)

num1 = 10
num2 = 5

num1 > num2
```

```python
## less than (<)

print(num1 < num2)
```

```python
##3 greater than or equal to (>=)

num1 = 30
num2 = 30

print(num1 >= num2)
```

```python
## less than or equal to (<=)

print(num1 <= num2)
```

### Logical Operators

```python
## And, Not, OR

x = True
y = True

result = x and y # returns True if both x and y are True, otherwise returns False
print(result)
```

```python
x = False
y = True

result = x and y # returns True if both x and y are True, otherwise returns False
print(result)
```

```python
## NOT

x = True

print(not x) # returns False because x is True
```

```python
## OR

x = True
y = False

result = x or y # returns True if at least one of x or y is True, otherwise returns False
print(result)
```

```python
x = False
y = False

result = x or y # returns True if at least one of x or y is True, otherwise returns False
print(result)
```

```python
x = True
y = True

result = x or y # returns True if at least one of x or y is True, otherwise returns False
print(result)
```
