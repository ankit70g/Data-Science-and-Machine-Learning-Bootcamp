# Variables

Variables are fundamental elements in programming used to store data that can be referenced and manipulated in a program. In Python, variables are created when you assign a value to them, and they do not need explicit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable.

### Topics:
- Introduction to Variables
- Declaring and assigning variables
- Naming conventions
- Understanding variable types
- Type checking and conversion
- Dynamic typing

```python
## Declaring and Assigning Variables

age=18
height=6.1
name="Ankit"
is_student=True

print("age:", age)
print("height:", height)
print("name:", name)
print("is_student:", is_student)
```

```python
## Naming Conventions
## Variables names should be descriptive
## They must start with a letter or an underscore '_'
## Variables names case sensitive

#valid variable names

first_name = "Ankit"
last_name = "Gupta"
```

```python
## Invalid variable names

#2age = 18
#first-name = "Ankit"
@name = "Ankit"
```

```python
## Case Sensitivity

name = "Ankit"
Name = "Gupta"
```

```python
##Understanding Variable Types
## Python is dynamically typed, type of a variable is determined at runtime

age = 18 # int
height = 6.1 # float
name = "Ankit" # str
is_student = True # bool

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

```python
## Type Checking and Type Conversion

type(height)
```

```python
age = 18
print(type(age))

#Type conversion
age_str = str(age) # This will convert the integer 18 to the string "18"
print(age_str)
print(type(age_str)) # This will show that age_str is now of type str
```

```python
age = "19"
print(type(int(age))) # This will convert the string "19" to the integer 19
```

```python
name = "Ankit"
int(name) # This will raise an error as "Ankit" cannot be converted to an integer
```

```python
height= 6.1
int(height) # This will convert the float 6.1 to the integer 6, truncating the decimal part
```

```python
float(int(height)) # This will first convert 6.1 to 6 and then back to 6.0 as a float
```

```python
## Dynamic Typing
## Python allows you to change the type of a variable by reassigning it

var = 10 # var is an integer
print(var, type(var))

var = "Hello" # var is now a string
print(var, type(var))

var = 3.14 # var is now a float
print(var, type(var))
```

```python
## input() function to take user input

age = int(input("What is the age")) # input() returns a string, we convert it to an integer using int()
print(age, type(age))
```

```python
## Simple Calculator

num1 = float(input("Enter first number:"))
num2 = float(input("Enter second number:"))

sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
