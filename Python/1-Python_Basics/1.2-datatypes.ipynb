# Data Types

1. Definition:
   - Data types are a classification of data which tell the compiler or interpreter how the programmer intends to use the data.
   - They determine the type of operations that can be performed on the data, the values that the data can take, and the amount of memory needed to store the data.

2. Importance of Data Types in Programming
   - Data types ensure that data is stored in an efficient way
   - They help in performing correct operations on data
   - Proper use of data types can prevent errors and bugs in the program

```python
## Integers
age = 18
type(age)
```

```python
## Floating-point data type
height = 6.1
print(height)
print(type(height))
```

```python
## String data type
name = "Ankit"
print(name)
print(type(name))
```

```python
#3 Boolean data type
is_adult = True
print(is_adult)
print(type(is_adult))
```

```python
a=10
b=10

type(a==b) # This will return <class 'bool'> because a==b evaluates to True, which is a boolean value
```

```python
## Common errors

result = "Hello" + 5 # This will raise a TypeError because you cannot concatenate a string and an integer
```

```python
result = "Hello" + str(5) # This will work because we are converting the integer 5 to a string before concatenating
print(result) # This will print "Hello5"
```
