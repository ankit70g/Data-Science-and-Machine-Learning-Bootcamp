```python
my_dict = {i: i**2 for i in range(1, 11)}
new_dict = {k: v for k, v in my_dict.items() if k % 2 == 0}

print(my_dict)
print(new_dict)
```

```python
my_dict = {i: i**2 for i in range(1, 6)}
new_dict = {v: k for k, v in my_dict.items()}

print(my_dict)
print(new_dict)
```

```python
from collections import defaultdict
dd = defaultdict(list)
dd['colors'].append('red')
dd['colors'].append('blue')

print(dict(dd))
```

```python
def count_chars(s):
    return {char: s.count(char) for char in set(s)}

print(count_chars("hello world"))
```
