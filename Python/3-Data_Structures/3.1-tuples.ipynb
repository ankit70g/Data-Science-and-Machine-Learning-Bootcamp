1. Introduction to Tuples

    Python Tuple is a collection of objects separated by commas. A tuple is similar to a Python list in terms of indexing, nested objects, and repetition but the main difference between both is Python tuple is immutable, unlike the Python list which is mutable.

2. Creating Tuples

```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))

tpl = tuple()
print(type(tpl))
```

Converting list into tuples

```python
numbers = tuple([1,2,3,4,5])
print(numbers)
```

Converting tuples into list

```python
lst = list((1,2,3,4,5))
print(lst)
```

```python
mixed_tuple = (1, "Hell0", 3.14, True)
print(mixed_tuple)
```

3. Accessing Tuple Elements

    Tuples in Python provide two ways by which we can access the elements of a tuple.

- Python Access Tuple using a Positive Index

```python
numbers = (1, 2, 3, 4)
print(numbers[2])
```

- Access Tuple using Negative Index


```python
print(numbers[-3])
```

```python
print(numbers[0:4])
```

```python
numbers[::-1]
```

4. Tuple Operations

- Concatenation of Python Tuples

```python
concatenation_tuple = numbers + mixed_tuple
print(concatenation_tuple)
```

- Repetition Python Tuples

```python
mixed_tuple * 3
```

```python
numbers * 3
```

- Traversing Items of Python Tuples

```python
tup = (1, 2, 3, 4, 5)

# Traverse through each item in the tuple
for x in tup:
    print(x, end=" ")
```

5. Tuples Slicing

```python
tup = (0 ,1, 2, 3)

print(tup[1:])
print(tup[::-1])
print(tup[2:4])
```

8. Packing and Unpacking Tuples

Packing - Packing means putting multiple values into a single tuple.

```python
packed_tuple = 1, "Hello", 3.14
print(packed_tuple)
```

Unpacking - Unpacking means extracting values from a tuple into variables.

```python
a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

Unpacking using * (Star Expression) — Advanced Unpacking

Used when elements are unknown or extra values exist.

```python
numbers = (1, 2, 3, 4, 5, 6)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

9. Nested Tuples

```python
lst = [[1,2,3,4], [6,7,8,9],[1,"Hello",3.14,"c"]]
lst[0][0:3]
```

```python
nested_tuple = ((1,2,3), ("a", "b", "c"), (True, False))

print(nested_tuple[0])
print(nested_tuple[1][2])
```

Iterating over nested tuples

```python
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
```
