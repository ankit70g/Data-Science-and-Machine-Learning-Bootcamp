1. Introduction to Dictionaries

    Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

2. Creating Dictionaries

    A dictionary is created by writing key-value pairs inside { }, where each key is connected to a value using colon (:). A dictionary can also be created using dict() function.

```python
empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
print(empty_dict)
```

```python
student = {"name": "Ankit", "age": "19", "grade": 20 }
print(student)
print(type(student))
```

```python
# Single key is always used
student = {"name": "Ankit", "age": 12, "name": 23}
print(student)
```

3. Accessing Dictionary Elements

    A value in a dictionary is accessed by using its key. This can be done either:
    - with square brackets [ ]
    - get() method. 
    Both return the value linked to the given key.






```python
student = {"name": "Ankit", "age": "19", "grade": "A" }
print(student)
```

```python
print(student["grade"])
print(student["age"])
```

```python
print(student.get("grade"))
print(student.get("last_name"))
```

        To give default value instead of None

```python
print(student.get("last_name", "Not Available"))
```

4. Modifying Dictionary Elements

    Dictionary are mutable, so you can add, update or delete elements


- Updating the dictionary
    - = (Assignment operator) is used to update the dictionary by giving a new key and a value.

    - If an existing key is used with the assignment operator, its value is updated with the new one.

```python
student["roll no"] = 22
print(student)
student["location"] = "Mumbai"
print(student)
```

- Deleting key value pair
    - del: removes an item using its key
    - pop(): removes the item with the given key and returns its value
    - clear(): removes all items from the dictionary
    - popitem(): removes and returns the last inserted key-value pair

```python
del student["location"]
print(student)
```

```python
val = student.pop("roll no")
print(val)
```

```python
print(student)
```

```python
my_dict = {"name": "Ankit", "roll no": 19, "age": 19, "location": "Mumbai", "id": 1234}

print(my_dict.popitem())
```

```python
print(my_dict.clear())
```

5. Dictionary Methods

    Python Dictionary is like a map that is used to store data in the form of a key: value pair.

keys() - Returns a view object that displays a list of all the keys in the dictionary in order of insertion

```python
students = {"name": "Ankit", "age": 21, "location": "Mumbai", "roll no": 19}
```

```python
keys = students.keys()
print(keys)
```

values() - Returns a view object containing all dictionary values, which can be accessed and iterated through efficiently

```python
values = students.values()
print(values)
```

setdefault() - Returns the value of a key if the key is in the dictionary else inserts the key with a value to the dictionary

```python
set_default = students.setdefault("location")
print(set_default)
```

```python
set_def = students.setdefault("emp_id", "4321")
print(set_def)
print(students)
```

items() - Return the list with all dictionary keys with values

```python
print(students.items())
```

fromkeys() - Creates a dictionary from the given sequence

```python
keys = students.fromkeys(students)
print(keys)
```

```python
def_value = students.fromkeys(students, "Not available")
print(def_value)
```

Using strings as keys

```python
data = students.fromkeys("age", 0)
print(data)
```

```python
print(students)
```

copy() - Returns a shallow copy of the dictionary

```python
student_copy = students
print(students)
print(student_copy)
```

```python
students["name"] = "Ankit2"
print(students)
print(student_copy)
```

```python
student_copy1 = students.copy()
print(student_copy1)
print(students)
```

```python
students["name"] = "Ankit3"
print(student_copy1)
print(students)
```

6. Iterating over Dictionaries

    A dictionary can be traversed using a for loop to access its keys, values or both key-value pairs by using the built-in methods keys(), values() and items().

```python
#Iterate over keys
for keys in students.keys():
    print(keys)
```

```python
#Iterate over values

for values in students.values():
    print(values)
```

```python
# Iterate over key-value pairs

for key, value in students.items():
    print(f"{key}: {value}")
```

7. Nested Dictionaries

    A nested dictionary is a dictionary that contains another dictionary as one of its values.

```python
student = {
    "student1": {"name": "Ankit", "age": 19},
    "student2": {"name": "Shanaya", "age": 20}
}

print(student)
```

```python
# Access nested dictionaries elements

print(student["student1"]["name"])
print(student["student2"]["age"])
```

```python
student.items()
```

Iterating over Nested Dictionaries

```python
for student_id, student_info in student.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info.items():
        print(f"{key}: {value}")
```

8. Dictionary Comprehension

    Dictionary comprehension is used to create a dictionary in a short and clear way. It allows keys and values to be generated from a loop in one line. This helps in building dictionaries directly without writing multiple statements.

```python
squares = {x:x**2 for x in range(10)}
print(squares)
```

```python
# Condition Dictionary Comprehension

evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
# Creatinf Dictionary from two lists

keys = {"a", "b", "c"}
values = {1, 2, 3}

data = {k:v for (k, v) in zip(keys, values)}
print(data)
```

9. Practical Examples

```python
# Use a dictionary to count the frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
# Merge 2 dictionaries into one

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged_dict = {**dict1, **dict2}
print(merged_dict)
```
