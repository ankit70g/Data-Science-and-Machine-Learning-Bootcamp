1. Introduction to Sets

    Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

3. Basics Sets Operations

Adding an element
- add() - We can add single element in a set
- update() - To add multiple items

```python
my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)

```

```python
my_set.update([8, 9])
print(my_set)
```

Removing Elements from the Set
- remove() - removes a specified element from the set

```python
my_set.remove(1)
print(my_set)
```

```python
my_set.remove(10)
print(my_set)
```

- discard() - removes a specified element from the set. Unlike remove(), if the element is not found, it does not raise an error.

```python
my_set.discard(11)
print(my_set)
```

- pop() - removes and returns an arbitrary element from the set. It uses FIFO method

```python
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

- clear() - removes all elements from the set, leaving it empty.

```python
my_set.clear()
print(my_set)
```

4. Sets Typecasting

    Typecasting objects refers to converting various data types into a set. Python provides the set() constructor to perform this typecasting, allowing us to convert lists, tuples and strings into sets.

```python
lst = [1, 2, 3, 4, 5]
s = set(lst)
print(s)
```

```python
d = {1: "One", 2: "Two", 3: "Three"}
s = set(d)
print(s)
```

5. Sets Membership Test

    Membership means checking whether an element exists inside a set or not.

- in → checks if element exists

```python
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)
print(10 in my_set)
```

- not in → checks if element does NOT exist

```python
my_set = {1, 2, 3, 4, 5, 6}
print(3 not in my_set)
print(10 not in my_set)
```

6. Mathematical Operations

    Python provides built-in operations for performing set operations such as union, intersection, difference and symmetric difference. In this article, we understand these operations one by one.

- union() - The union of two sets combines all unique elements from both sets.
    - set1.union(set2)  # Using the union() method
    - set1 | set2  # Using the '|' operator

```python
set1 = {1, 2, 3, 4, 5, 6, 7}
set2 = {"Ankit", "Shanaya", "Ritika"}

union_set1 = set1.union(set2)
union_set2 = set1 | set2
print(union_set1)
print(union_set2)
```

- intersection() - The intersection of two sets includes only the common elements present in both sets.
    - set1 & set2  # Using the '&' operator
    - set1.intersection(set2)  # Using the intersection() method

```python
set1 = {1, 2, 3, 4, 5}
set2 = {2, 4, 1, 7, 8, 9}
intersection_set1 = set1 & set2
intersection_set2 = set1.intersection(set2)
print(intersection_set1)
print(intersection_set2)
```

intersection_update() → changes the existing set

```python
set1.intersection_update(set2)
print(set1)
```

- difference() - The difference between two sets includes elements present in the first set but not in the second.
    - set1 - set2  # Using the '-' operator
    - set1.difference(set2)  # Using the difference() method

```python
set1 = {1, 2, 3, 4, 5}
set2 = {3, 5, 7, 9}

difference_set1 = set1 - set2
difference_set2 = set1.difference(set2)
print(difference_set1)
print(difference_set2)
```

difference_update() removes common elements from the original set

```python
set1.difference_update(set2)
print(set1)
```

- Symmetric Difference of sets

    The symmetric difference of two sets includes elements that are in either set but not in both.

    Syntax:
    - set1 ^ set2  # Using the '^' operator
    - set1.symmetric_difference(set2)# Using the symmetric_difference() method

```python
set1 = {1, 2, 3, 4, 5}
set2 = {6, 7, 8, 2, 1, 3}

symmetric_set1 = set1 ^ set2
symmetric_set2 = set1.symmetric_difference(set2)

print(symmetric_set1)
print(symmetric_set2)
```

7. Set Methods

- issubset() - Checks whether all elements of one set are present inside another set.

```python
set1 = {1, 2, 3}
set2 = {4, 5, 6}

sub_set = set1.issubset(set2)
print(sub_set)
```

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5, 6}

sub_set = set1.issubset(set2)
print(sub_set)
```

- superset() - Checks whether a set contains all elements of another set.

```python
super_set = set1.issuperset(set2)
print(super_set)
```

- isdisjoint() - Checks whether two sets have NO common elements.

```python
dis_set = set1.isdisjoint(set2)
print(dis_set)
```

```python
set1 = {1, 2, 3}
set2 = {4, 5, 6}

dis_set = set1.isdisjoint(set2)
print(dis_set)
```
