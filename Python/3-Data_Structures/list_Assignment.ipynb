```python
import random

random_numbers = [random.randint(1, 20) for _ in range(10)]
print("Original random list:", random_numbers)

#sort ascending
random_numbers.sort()
print(random_numbers)

#sort descending
random_numbers.sort(reverse=True)
print(random_numbers)

#remove duplicates
unique_numbers = list(set(random_numbers))
print(unique_numbers)
```

```python
students = [
    {"name": "Ankit", "score": 85},
    {"name": "Rishita", "score": 90},
    {"name": "Kirti", "score": 95}
]

#sort by score descending
sorted_students = sorted(students, key = lambda x: x["score"], reverse=True)
print(sorted_students)
```

```python
def transpose(matrix):
    return [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]

matrix_3x3 = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposed_matrix = transpose(matrix_3x3)
print(matrix_3x3)
print(transposed_matrix)
```

```python
def flatten(nested_list):
    return [element for sublist in nested_list for element in sublist]

nested_list = [[1,2,3],[4,5],[6,7,8]]
flattened_list = flatten(nested_list)
print(nested_list)
print(flattened_list)
```

```python
#Create a list of the first 10 positive integers. Remove the elements at indices 2, 4, and 6, and insert the element '99' at index 5. Print the modified list.
lst = list(range(1, 11))

for index in sorted([2, 4, 6], reverse=True):
    lst.pop(index)

#insert 99 at index 5
lst.insert(5, 99)
print(lst)
```

```python
list1 = [1, 2, 3, 4]
list2 = ['a', 'b', 'c', 'd']

zipped_list = list(zip(list1, list2))
print(zipped_list)
```

```python
def reverse_list(lst):
    return lst[::-1]

original_list = [1, 2, 3, 4, 5]
reversed_list = reverse_list(original_list)
print(original_list)
print(reversed_list)
```

```python
def rotate_list(lst, n):
    n = n % len(lst)
    return lst[-n:] + lst[:-n]

original_list = [1, 2, 3, 4, 5]
rotated_list = rotate_list(original_list, 2)
print(original_list)
print(rotated_list)
```

```python
def list_intersection(list1, list2):
    return [i for i in list1 if i in list2]

list2 = [1, 2 ,3, 4, 5]
list2 = [3, 4, 5, 6]

intersected_list = list_intersection(list1, list2)
print(intersected_list)
```
