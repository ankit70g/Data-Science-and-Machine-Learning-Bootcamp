```python
tpl = tuple(range(1, 11))
print(tpl)
```

```python
first_element = tpl[0]
middle_element = tpl[len(tpl) // 2]
last_element = tpl[-1]

print(first_element)
print(middle_element)
print(last_element)
```

```python
first_three_element = tpl[:3]
last_three_element = tpl[-3]
specific_index = tpl[2:6]

print(first_three_element)
print(last_three_element)
print(specific_index)
```

```python
def tuple_min(t):
    return min(t)

def tuple_max(t):
    return max(t)

def tuple_sum(t):
    return sum(t)

sample = (3, 7, 1, 9, 4, 6, 2)
print(sample)
print(tuple_min(sample))
print(tuple_max(sample))
print(tuple_sum(sample))
```
