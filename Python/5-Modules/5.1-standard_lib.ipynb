5. Data Serialization

    Data serialization is the process of converting a data structure or object into a format that can be stored (like a file) or transmitted (over a network) and then reconstructed later. In Python, this is essential for saving program states or sending data between different applications.

<div style="margin-left:40px">
a. JSON (JavaScript Object Notation)

JSON is the most common format for serialization because it is human-readable and language-independent. It’s perfect for web APIs and configuration files.
- Library: json (Built-in)
- Best for: Cross-platform communication.

```python
import json

data = {"name": "Ankit", "age": 19}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

<div style="margin-left:40px">
b. Pickle

Pickle is Python’s native serialization module. Unlike JSON, it can serialize almost any Python object (like custom classes or complex functions).
- Library: pickle (Built-in)
- Warning: Never "unpickle" data from an untrusted source, as it can execute arbitrary code.
- Best for: Saving complex Python-specific objects for later use in Python.

```python
import pickle

data = [1, 2, 3, {"a": 1}]

# Serialize: Write to a binary file
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)

# Deserialize: Read from binary file
with open("data.pkl", "rb") as f:
    loaded_data = pickle.load(f)
```

<div style="margin-left:40px">
c. CSV (Comma Separated Values)

While not "object serialization" in the strict sense, CSV is the standard for serializing tabular data, especially in data science.
- Library: csv or pandas
- Best for: Spreadsheets and large datasets.

```python
import csv

# Simple serialization to CSV
with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Score"])
    writer.writerow(["Ankit", 95])

with open('example.csv', mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```
