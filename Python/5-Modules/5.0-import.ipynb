1. Introduction to modules and packages

    In Python, modules and packages help organize and reuse code. Here is a comprehensive guide on how to import them


2. Key Components of a python package

    - Module: A single Python file containing reusable code (e.g., math.py).
    - Package: A directory containing modules and a special __init__.py file.
    - Sub-Packages: Packages nested within other packages for deeper organization.


3. Importing packages and modules

<div style="margin-left:40px">
a. The Standard Import

This is the most basic way to import an entire module. When you use this method, you have to prefix the functions or classes you use with the module's name.

```python
import math
math.sqrt(16)
```

<div style="margin-left:40px">
b. Importing Specific Parts (from ... import)

If you only need a specific function, class, or variable from a module, you can import just that part. This allows you to use the function directly without typing the module name first.

```python
from math import sqrt,pi
print(sqrt(25))
print(sqrt(16))
print(pi)
```

<div style="margin-left:40px">
c. Importing with an Alias (as)

If a module name is long or you want to save keystrokes, you can give it a temporary nickname using the as keyword. This is very common with popular data science libraries.

```python
import numpy as np
np.array([1,2,3,4,5])
```

<div style="margin-left:40px">
d. Importing Specific Parts with an Alias

You can also assign an alias to a specific function or class if its name conflicts with something else in your code, or if it's just too long.

```python
from datetime import datetime as dt

current_time = dt.now()
print(current_time)
```

<div style="margin-left:40px">
e. Importing Everything (*)

You can import all functions and classes from a module using the asterisk.

    Note: This is generally considered bad practice in large projects because it clutters your namespace and makes it hard to tell which module a specific function came from.

```python
from math import * 
print(sqrt(16))
print(pi)
```

4. Creating and Acessing packages

    - Create a Directory: create a folder that will act as the package root.
    - Add Modules: Inside the directory, add Python files (modules). Each module can contain related functions or classes.
    - Add __init__.py: add an __init__.py file to the directory. This file tells Python that the directory should be treated as a package.
    - Create Sub-packages (Optional): One can organize code further by creating subdirectories with their own __init__.py files.
    - Import Modules: Modules or functions inside the package can be imported using dot notation, for example:

        from mypackage.module1 import greet

```python
from package.maths import addition
addition(2, 6)
```

```python
from package import maths
maths.addition(2, 8)
```
